```python
def solution1(strs):
    if not strs:
        return ""

    prefix = strs[0]
    for word in strs[1:]:
        common_length = 0
        limit = min(len(prefix), len(word))
        while common_length < limit and prefix[common_length] == word[common_length]:
            common_length += 1
        prefix = prefix[:common_length]
        if not prefix:
            break
    return prefix


def solution2(arr, target):
    left, right = 0, len(arr)
    while left < right:
        middle = (left + right) // 2
        if arr[middle] < target:
            left = middle + 1
        else:
            right = middle
    return left


if __name__ == "__main__":
    print(solution1(["flower", "flow", "flight"]))
    print(solution1(["car", "star", "wars"]))

    for values, target in [
        ([1, 2, 3, 6, 9], 9),
        ([1, 2, 3, 6, 9], 5),
        ([1, 2, 3, 6, 9], 0),
        ([1, 2, 3, 6, 9], 110),
    ]:
        print(solution2(values, target))
```

- Общий префикс: время `O(Σ min(|s₀|, |sᵢ|))`, в худшем случае `O(n·L)`; память `O(L)` для префикса и результата.
- Позиция вставки: время `O(log n)`, дополнительная память `O(1)`.
